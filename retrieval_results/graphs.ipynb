```python
import json
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def get_json(json_file): 
    with open(json_file, "r") as f:
        return json.load(f)

def get_avg(results, metric):
    records = []

    for res in results:
        met = res.get(metric)

        if met is None:
            continue

        records.append({
            "topk": len(res.get("hits")),
            "embedder": res.get("embedder"),
            metric: round(float(met), 2),
        })

    df = pd.DataFrame(records)
    # A few queries can come back with fewer hits than requested (corpus too small
    # to fill k), which would otherwise leak a spurious extra topk column here.
    dominant_topk = df["topk"].mode().iloc[0]
    df = df[df["topk"] == dominant_topk]
    return df.pivot_table(index="embedder", columns="topk", values=metric, aggfunc="mean")

dense_topk_is_3 = get_json("retrieval_20260725_154313.json")
dense_topk_is_5 = get_json("retrieval_20260724_193108.json")
dense_topk_is_7 = get_json("retrieval_20260725_184058.json")
dense_topk_is_9 = get_json("retrieval_20260727_081522.json")
dense_topk_is_11 = get_json("retrieval_20260727_081524.json")

sparse_topk_is_3 = get_json("retrieval_20260725_154255.json")
sparse_topk_is_5 = get_json("retrieval_20260725_115511.json")
sparse_topk_is_7 = get_json("retrieval_20260725_154244.json")
sparse_topk_is_9 = get_json("retrieval_20260727_081547.json")
sparse_topk_is_11 = get_json("retrieval_20260727_081612.json")
```

```python
dense_topk_is_3_avg_prec = get_avg(dense_topk_is_3, "context_precision")
dense_topk_is_5_avg_prec = get_avg(dense_topk_is_5, "context_precision")
dense_topk_is_7_avg_prec = get_avg(dense_topk_is_7, "context_precision")
dense_topk_is_9_avg_prec = get_avg(dense_topk_is_9, "context_precision")
dense_topk_is_11_avg_prec = get_avg(dense_topk_is_11, "context_precision")

dense_topk_is_3_avg_rec = get_avg(dense_topk_is_3, "context_recall")
dense_topk_is_5_avg_rec = get_avg(dense_topk_is_5, "context_recall")
dense_topk_is_7_avg_rec = get_avg(dense_topk_is_7, "context_recall")
dense_topk_is_9_avg_rec = get_avg(dense_topk_is_9, "context_recall")
dense_topk_is_11_avg_rec = get_avg(dense_topk_is_11, "context_recall")

sparse_topk_is_3_avg_prec = get_avg(sparse_topk_is_3, "context_precision")
sparse_topk_is_5_avg_prec = get_avg(sparse_topk_is_5, "context_precision")
sparse_topk_is_7_avg_prec = get_avg(sparse_topk_is_7, "context_precision")
sparse_topk_is_9_avg_prec = get_avg(sparse_topk_is_9, "context_precision")
sparse_topk_is_11_avg_prec = get_avg(sparse_topk_is_11, "context_precision")

sparse_topk_is_3_avg_rec = get_avg(sparse_topk_is_3, "context_recall")
sparse_topk_is_5_avg_rec = get_avg(sparse_topk_is_5, "context_recall")
sparse_topk_is_7_avg_rec = get_avg(sparse_topk_is_7, "context_recall")
sparse_topk_is_9_avg_rec = get_avg(sparse_topk_is_9, "context_recall")
sparse_topk_is_11_avg_rec = get_avg(sparse_topk_is_11, "context_recall")
```

```python
precision_df = pd.concat([
    pd.concat([dense_topk_is_3_avg_prec, dense_topk_is_5_avg_prec, dense_topk_is_7_avg_prec, dense_topk_is_9_avg_prec, dense_topk_is_11_avg_prec], axis=1),
    pd.concat([sparse_topk_is_3_avg_prec, sparse_topk_is_5_avg_prec, sparse_topk_is_7_avg_prec, sparse_topk_is_9_avg_prec, sparse_topk_is_11_avg_prec], axis=1),
])

recall_df = pd.concat([
    pd.concat([dense_topk_is_3_avg_rec, dense_topk_is_5_avg_rec, dense_topk_is_7_avg_rec, dense_topk_is_9_avg_rec, dense_topk_is_11_avg_rec], axis=1),
    pd.concat([sparse_topk_is_3_avg_rec, sparse_topk_is_5_avg_rec, sparse_topk_is_7_avg_rec, sparse_topk_is_9_avg_rec, sparse_topk_is_11_avg_rec], axis=1),
])

combined_df = pd.concat({"context_precision": precision_df, "context_recall": recall_df}, axis=1)
combined_df
```

```python
EMBEDDER_LABELS = {
    "text-embedding-3-small": "te3-small",
    "Qwen3-Embedding": "Qwen3",
    "bert-base-multilingual-cased": "mBERT",
    "SlovakBERT": "SlovakBERT",
    "xlm-roberta-base": "XLM-R",
    "czert-bert-base-cased": "Czert",
    "BM25": "BM25",
}
METRIC_TITLES = {
    "context_precision": "Context precision",
    "context_recall": "Context recall",
}

def to_latex_topk_table(combined_df, ks=(3, 5, 7, 9, 11),
                         caption="Effect of retrieval depth $k$ on mean RAGAS context precision and recall, "
                                 "aggregated across all repositories. BM25 is the sparse lexical baseline. "
                                 "Best value per column in bold.",
                         label="tab:topk"):
    prec = combined_df["context_precision"]
    dense_embedders = [e for e in prec.index if e != "BM25"]
    dense_order = prec.loc[dense_embedders, list(ks)].mean(axis=1).sort_values(ascending=False).index.tolist()

    lines = [
        r"\begin{table}[H]",
        r"\centering",
        r"\caption{%s}" % caption,
        r"\label{%s}" % label,
        r"\scriptsize",
        r"\setlength{\tabcolsep}{5pt}",
        r"\begin{tabular}{l" + "c" * len(ks) + "}",
        r"\hline",
        r"\textbf{Embedder} & " + " & ".join(r"\textbf{$k{=}%d$}" % k for k in ks) + r" \\",
        r"\hline",
    ]

    for metric in ("context_precision", "context_recall"):
        sub = combined_df[metric]
        col_max = sub.max(axis=0)
        lines.append(r"\multicolumn{%d}{c}{\textit{%s}} \\" % (len(ks) + 1, METRIC_TITLES[metric]))
        lines.append(r"\hline")

        for e in dense_order:
            vals = []
            for k in ks:
                v = sub.loc[e, k]
                s = f"{v:.2f}"
                if abs(v - col_max[k]) < 1e-9:
                    s = r"\textbf{%s}" % s
                vals.append(s)
            lines.append(f"{EMBEDDER_LABELS.get(e, e)} & " + " & ".join(vals) + r" \\")
        lines.append(r"\hline")

        vals = []
        for k in ks:
            v = sub.loc["BM25", k]
            s = f"{v:.2f}"
            if abs(v - col_max[k]) < 1e-9:
                s = r"\textbf{%s}" % s
            vals.append(s)
        lines.append(r"BM25 \textit{(sparse)} & " + " & ".join(vals) + r" \\")
        lines.append(r"\hline")

    lines.append(r"\end{tabular}")
    lines.append(r"\end{table}")
    return "\n".join(lines)

print(to_latex_topk_table(combined_df))

with open("table.tex", "w") as f:
    f.write(to_latex_topk_table(combined_df))
```

```python
import sys
from pathlib import Path

import numpy as np
from scipy.stats import false_discovery_control

sys.path.append("..")

from wilcoxon_dense_vs_baseline import load_dense_scores, paired_samples, EMBEDDERS as DENSE_EMBEDDERS
from wilcoxon_sparse_vs_baseline import load_sparse_scores, ARM_SUFFIXES
from scipy.stats import wilcoxon

METRICS = ["context_precision", "context_recall"]


def add_fdr_correction(df, alpha=0.05):
    """Benjamini-Hochberg FDR correction over all tests in df (one family)."""
    df = df.copy()
    mask = df["p_value"].notna()
    adjusted = np.full(len(df), np.nan)
    adjusted[mask.to_numpy()] = false_discovery_control(df.loc[mask, "p_value"].to_numpy(), method="bh")
    df["p_value_adj"] = adjusted
    df["significant_adj"] = df["p_value_adj"] <= alpha
    return df
```

```python
# Dense: is k=3, k=7, k=9, or k=11 significantly different from the k=5 baseline, per embedder?
DENSE_BASELINE_FILE = "retrieval_20260724_193108.json"  # k=5
DENSE_COMPARISON_FILES = {
    3: "retrieval_20260725_154313.json",
    7: "retrieval_20260725_184058.json",
    9: "retrieval_20260727_081522.json",
    11: "retrieval_20260727_081524.json",
}

baseline_scores = load_dense_scores(Path(DENSE_BASELINE_FILE))

rows = []
for k, file in DENSE_COMPARISON_FILES.items():
    comp_scores = load_dense_scores(Path(file))
    for metric in METRICS:
        for embedder in DENSE_EMBEDDERS:
            a, b = paired_samples(comp_scores[embedder][metric], baseline_scores[embedder][metric])
            if not a or all(x == y for x, y in zip(a, b)):
                rows.append({"embedder": embedder, "metric": metric, "k": k, "n": len(a),
                             "statistic": None, "p_value": None, "significant": False})
                continue
            result = wilcoxon(a, b, zero_method="wilcox")
            rows.append({"embedder": embedder, "metric": metric, "k": k, "n": len(a),
                         "statistic": result.statistic, "p_value": result.pvalue,
                         "significant": result.pvalue <= 0.05})

dense_sig_df = add_fdr_correction(pd.DataFrame(rows))
dense_sig_df
```

```python
# Sparse: is k=3, k=7, k=9, or k=11 significantly different from the k=5 baseline, per arm (BM25 / SPLADE)?
# Both arms are mislabeled embedder="BM25" in these files, so they're split by collection suffix.
SPARSE_BASELINE_FILE = "retrieval_20260725_115511.json"  # k=5
SPARSE_COMPARISON_FILES = {
    3: "retrieval_20260725_154255.json",
    7: "retrieval_20260725_154244.json",
    9: "retrieval_20260727_081547.json",
    11: "retrieval_20260727_081612.json",
}

rows = []
for k, file in SPARSE_COMPARISON_FILES.items():
    for metric in METRICS:
        for arm, suffix in ARM_SUFFIXES.items():
            baseline_arm_scores = load_sparse_scores(Path(SPARSE_BASELINE_FILE), suffix)
            comp_arm_scores = load_sparse_scores(Path(file), suffix)
            a, b = paired_samples(comp_arm_scores[metric], baseline_arm_scores[metric])
            if not a or all(x == y for x, y in zip(a, b)):
                rows.append({"arm": arm, "metric": metric, "k": k, "n": len(a),
                             "statistic": None, "p_value": None, "significant": False})
                continue
            result = wilcoxon(a, b, zero_method="wilcox")
            rows.append({"arm": arm, "metric": metric, "k": k, "n": len(a),
                         "statistic": result.statistic, "p_value": result.pvalue,
                         "significant": result.pvalue <= 0.05})

sparse_sig_df = add_fdr_correction(pd.DataFrame(rows))
sparse_sig_df
```

```python
wilcoxon_pvalues_df = pd.concat([
    dense_sig_df.rename(columns={"embedder": "group"}).assign(mode="dense"),
    sparse_sig_df.rename(columns={"arm": "group"}).assign(mode="sparse"),
], ignore_index=True)[["mode", "group", "metric", "k", "n", "statistic", "p_value", "significant", "p_value_adj", "significant_adj"]]

wilcoxon_pvalues_df.to_csv("wilcoxon_pvalues.csv", index=False)
wilcoxon_pvalues_df
```
